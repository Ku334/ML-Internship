```python
#py list vs np array
nums = [1, 2, 3, 4, 5]
print("List * 2:   ", nums * 2)               # repeats the list
print("Array * 2:  ", np.array(nums) * 2)     # doubles each value
```

Indexing and Slicing

```python
#1d indexing and slicing
a = np.array([10, 20, 30, 40, 50, 60, 70])
print("a[0]    :", a[0])      # first element
print("a[-1]   :", a[-1])     # last element
print("a[2:5]  :", a[2:5])    # index 2 up to (not including) 5
print("a[:3]   :", a[:3])     # first three
print("a[::2]  :", a[::2])    # every second element
print("a[::-1] :", a[::-1])   # reversed
```

```python
print("Values > 30        :", a[a > 30])
print("Values 20 < x < 60 :", a[(a > 20) & (a < 60)])
print("Even values        :", a[a % 2 == 0])
```

```python
#2d indexing and slicing
m = np.arange(1, 13).reshape(3, 4)
print(m)
```

```python
print( m[1, 2])     # row 1, column 2
print( m[0])        # first row
print( m[:, 1])     # second column
print( m[1:, :2])   # rows 1 onward, first two columns
```

```python
print("Last row   :", m[-1])
print("Last column:", m[:, -1])
```

Broadcasting

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print("Shape:", m.shape)

# Scalar broadcasting: one number is applied to every element
print(m + 100)

# Row broadcasting: shape (4,) is added to each row of the (3, 4) array
row = np.array([10, 20, 30, 40])
print(m + row)

# Column broadcasting: shape (3, 1) is added to each column of the (3, 4) array
col = np.array([[100], [200], [300]])
print("Column shape:", col.shape)
print(m + col)
```

Basic Statistics

```python
scores = np.array([72, 88, 95, 60, 79, 91])
print("Sum    :", scores.sum())
print("Mean   :", scores.mean())
print("Median :", np.median(scores))
print("Std    :", round(scores.std(), 2))
print("Min    :", scores.min())
print("Max    :", scores.max())
```

```python
marks = np.array([
    [80, 90, 70],
    [60, 75, 85],
    [95, 65, 88],
    [50, 70, 90],
])
print(marks)
print("Shape:", marks.shape)
```

```python
print("Overall mean         :", marks.mean())
print("Mean per column      :", marks.mean(axis=0))   
print("Mean per row         :", marks.mean(axis=1))   
print("Max per column       :", marks.max(axis=0))
print("Total per student    :", marks.sum(axis=1))
```

```python
# Sorting and percentiles
print("Sorted      :", np.sort(scores))
print("75th percentile:", np.percentile(scores, 75))
```
