Missing Values and Duplicates

```python
df = pd.read_csv(RAW_PATH)

missing = df.isnull().sum()
pd.DataFrame({
    "missing": missing,
    "percent": (missing / len(df) * 100).round(1),
}).query("missing > 0")
```

Duplicates

```python
print("Exact duplicate rows:", df.duplicated().sum())
df[df.duplicated(keep=False)].sort_values("student_id")
```

```python
print("student_id unique now:", df["student_id"].is_unique)
```

```python
before = len(df)
df = df.drop_duplicates()
print(f"Rows: {before} -> {len(df)}")
print("student_id unique now:", df["student_id"].is_unique)
```

```python
clean = df["gender"].str.strip().str.lower()
clean.value_counts()
```

```python
clean = df["gender"].str.strip().str.lower()

df["gender"] = np.select(
    [clean.isin(["m", "male"]), clean.isin(["f", "female"])],
    ["Male", "Female"],
    default="Unknown",
)

print(df["gender"].value_counts())
print("Unknown values:", (df["gender"] == "Unknown").sum())
```

```python
print("Before:", df["attendance_pct"].dtype, "|", df["attendance_pct"].dropna().head(4).tolist())

df["attendance_pct"] = pd.to_numeric(df["attendance_pct"].str.replace("%", "", regex=False))

print("After :", df["attendance_pct"].dtype, "|", df["attendance_pct"].dropna().head(4).tolist())
print("Mean attendance:", round(df["attendance_pct"].mean(), 1))
```

Impossible values

```python
print("Age above 70      :", (df["age"] > 70).sum())
print("Negative study hrs:", (df["study_hours"] < 0).sum())
print("Score above 100   :", (df["score"] > 100).sum())
```

```python
# Replace impossible values with NaN. We cannot know the true value, so we do not guess.
df["age"] = df["age"].mask(df["age"] > 70)
df["study_hours"] = df["study_hours"].mask(df["study_hours"] < 0)
df["score"] = df["score"].mask(df["score"] > 100)

print(df[["age", "study_hours", "score"]].describe().round(1))
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df = df.dropna(subset=["score"])

for col in ["age", "study_hours", "sleep_hours", "attendance_pct"]:
    df[col] = df[col].fillna(df[col].median())

df["age"] = df["age"].astype(int)
df = df.reset_index(drop=True)

print("Missing values left:", int(df.isnull().sum().sum()))
print("Shape:", df.shape)
df.dtypes
```

GroupBy

```python
print(df.groupby("gender")["score"].mean().round(1))
print()
print(df.groupby("city")["score"].mean().round(1).sort_values(ascending=False))
```

```python
city_summary = df.groupby("city").agg(
    students=("student_id", "count"),
    avg_score=("score", "mean"),
    avg_study=("study_hours", "mean"),
    avg_attendance=("attendance_pct", "mean"),
).round(1)
city_summary.sort_values("avg_score", ascending=False)
```

Merge

```python
provinces = pd.DataFrame({
    "city": ["Kathmandu", "Lalitpur", "Bhaktapur", "Pokhara", "Butwal", "Biratnagar", "Dharan", "Janakpur"],
    "province": ["Bagmati", "Bagmati", "Bagmati", "Gandaki", "Lumbini", "Koshi", "Koshi", "Madhesh"],
})
provinces
```

```python
left = pd.merge(df, provinces, on="city", how="left")     # keep every student
inner = pd.merge(df, provinces, on="city", how="inner")   # only rows with a match
outer = pd.merge(df, provinces, on="city", how="outer", indicator=True)   # keep everything

print("Students     :", len(df))
print("Left rows    :", len(left))
print("Inner rows   :", len(inner))
print("Outer rows   :", len(outer))
print()
print(outer["_merge"].value_counts())
```
